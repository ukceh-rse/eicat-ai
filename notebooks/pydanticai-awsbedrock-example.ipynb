```python
from pydantic_ai import Agent, DocumentUrl, ModelSettings
from IPython.display import Markdown, display
from pathlib import Path
from pydantic_ai import BinaryContent
from pydantic import BaseModel
from typing import List
```

```python
async def summarize(agent: Agent):
	result = await agent.run(
		[
			"Summarise the following document in less than 150 words. Structure your summary using markdown.",
			DocumentUrl(
				"https://portals.iucn.org/library/sites/library/files/documents/2020-026-En.pdf"
			),
		]
	)
	return result.output
```

```python
class Paper(BaseModel):
	abstract: str
	content: str
	references: List[str]

async def pdf_to_markdown(pdf_path: str):
    agent: Agent = Agent(
        "bedrock:anthropic.claude-3-7-sonnet-20250219-v1:0",
        output_type=Paper,
        model_settings=ModelSettings(max_tokens=10_000, temperature=0.0),
    )
    p: Path = Path(pdf_path)
    result = await agent.run(
        [
            "Convert the following pdf file into markdown format.",
            BinaryContent(data=p.read_bytes(), media_type="application/pdf"),
        ]
    )
    return result
```

```python
result = await pdf_to_markdown("../data/Caiola de Sostoa 2005.pdf")
```
