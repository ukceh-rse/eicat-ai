```python
import requests
from pathlib import Path

def convert_tei_to_html(tei_file_path, api_url="http://localhost:8000/tei2html"):
    """
    Convert TEI file to HTML using the stylesheets API
    """
    tei_path = Path(tei_file_path)
    
    if not tei_path.exists():
        raise FileNotFoundError(f"TEI file not found: {tei_file_path}")
    
    with open(tei_path, 'rb') as f:
        files = {'file': f}
        response = requests.post(api_url, files=files)
    
    if response.status_code == 200:
        # Save HTML to same directory with .html extension
        html_path = tei_path.with_suffix('.html')
        with open(html_path, 'w', encoding='utf-8') as f:
            f.write(response.text)
        print(f"HTML saved to: {html_path}")
        return str(html_path)
    else:
        raise Exception(f"API request failed: {response.status_code} - {response.text}")

# Example usage:
# html_file = convert_tei_to_html("/path/to/your/file.tei.xml")
```

```python
from html_to_markdown import convert

html_path = "/home/mpc/eicat-ai/data/eval/Barilani/Detecting_hybridization_in_wild.grobid.tei.html"
with (
    open(html_path, "r", encoding="utf-8") as input,
    open(f"{html_path}.md", "w") as output,
):
    html = input.read()
    md = convert(html)
    output.write(md)
```
